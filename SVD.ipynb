```python
def qr(X):
    m=len(X)
    n=len(X[0])
    R=np.eye(n)
    Q=[]
    for i in range(m):
        row=[]
        for j in range(n):
            row.append(0.0)
        Q.append(row)

    for j in range(n):
        v=[]
        for i in range(m):
            v.append(X[i][j])
        for k in range(j):
            dot=0
            for i in range(m):
                dot+=Q[i][k]*X[i][j]
            for i in range(m):
                v[i]-=dot*Q[i][k]
        l=0
        for i in range(m):
            l+=v[i]*v[i]
        l=l**0.5
        if abs(l)<1e-9:
            raise ValueError("can't divide by 0")
        for i in range(m):
            Q[i][j]=v[i]/l

    Q=np.array(Q)
    X=np.array(X)
    R=Q.T@X
    return Q,R
```

```python
def eig(x,itr=100):
    n=len(x)
    v=np.eye(n)
    x=np.array(x)
    X=x.astype(float).copy()

    for _ in range(itr):
        Q,R=qr(X)
        X=R@Q
        v=v@Q
    return np.diag(X),v
```

```python
def svd(A):
    A = np.array(A, dtype=float)
    m, n = A.shape

    # 1. Form B = A transpose A
    B = A.T @ A

    # 2. Find eigenvalues and eigenvectors
    raw_evals, eigvecs = eig(B)
    raw_evals = np.clip(raw_evals, 0, None)

    # 3. Sort eigenvalues and eigenvectors together
    sorted_indices = np.argsort(raw_evals)[::-1]
    evals = raw_evals[sorted_indices]
    V = eigvecs[:, sorted_indices]

    # 4. Singular values
    singular_values = np.sqrt(evals)

    sigma = np.zeros( ( min(m, n), min(m, n) ) )
    for i in range(min(m, n)):
        sigma[i, i] = singular_values[i]

    # 5. Calculate U columns
    U = np.zeros((m, n))

    for i in range(n):
        if singular_values[i] > 1e-8:
            U[:, i] = (A @ V[:, i]) / singular_values[i]

    return U, sigma, V.T
    
```
