```python
import numpy as np
```

```python
def d(matrix):
    n=len(matrix)
    if(n==1):
        return matrix[0][0]
    elif(n==2):
        return  matrix[0][0]*matrix[1][1] - matrix[0][1]*matrix[1][0]
        det=0
    else :
        for j in range(n):
            sign=(-1)**j
            
            minor=[]
            for i in range(1,n):
                row=[]
                for k in range(n):
                    if(k!=j):
                        row.append(matrix[i][k])
                minor.append(row)

                det+=sign*matrix[0][j]*d(minor)
```

```python
def adjoint(matrix):
    n = len(matrix)

    # Special case: 1x1 matrix
    if n == 1:
        return [[1]]

    cofactor = []

    # Calculate the cofactor matrix
    for i in range(n):
        row = []

        for j in range(n):
            minor = []

            # Remove row i and column j
            for r in range(n):
                if r == i:
                    continue

                minor_row = []

                for c in range(n):
                    if c != j:
                        minor_row.append(matrix[r][c])

                minor.append(minor_row)

            # Calculate cofactor
            value = ((-1) ** (i + j)) * d(minor)
            row.append(value)

        cofactor.append(row)

    # Transpose the cofactor matrix
    result = []

    for j in range(n):
        row = []

        for i in range(n):
            row.append(cofactor[i][j])

        result.append(row)

    return result

```

```python
#inverse of a matrix
```

```python
def inv(matrix):
    det=d(matrix)
    if det==0:
        print("the inverse doesn't exist")
    adj=adjoint(matrix)
    n=len(matrix)
    ans=[]
    for i in range(n):
        r=[]
        for j in range(n):
            value=adj[i][j]/det
            r.append(value)
        ans.append(r)
    return ans
```

```python
#row reduced echelon form
```

```python
def rref(matrix):
    A=[]
    for row in matrix:
        row=[float(x) for x in row]
        A.append(row)

    row=len(A)
    col=len(A[0])

    r=0
    for c in range(col):
        if r==row:
            break
        if A[r][c]==0:
            for i in range(r+1,row):
                if A[i][c]!=0:
                    A[r],A[i]=A[i],A[r]
                    break
        if A[r][c]==0:
            continue

        pivot=A[r][c]

        for j in range(col):
            A[r][j]=A[r][j]/pivot

        for i in range(row):
            if i!=r:
                factor=A[i][c]
                for j in range(col):
                    A[i][j]-=factor*A[r][j]

        r+=1
    return A
                
```

```python
A = [
    [1, 2, 3],
    [2, 4, 6],
    [1, 1, 1]
]

result = rref(A)

for row in result:
    print(row)
```

```python
#rank of  a matrix
```

```python
def rank(matrix):
    A=rref(matrix)
    r=0
    for i in range(len(A)):
        flag=False
        for j in range(len(A[0])):
            if(abs(A[i][j]))>1e-9:
                flag=True
        if flag==True:
            r+=1
    return r

A = [
    [1, 2, 3],
    [2, 4, 6],
    [1, 1, 1]
]

result = rank(rref(A))
print(result)
```

```python
#conversion to augmented matrix 
def solve(A,B):
    M=[]
    for i in range(len(A)):
        row=[]
        for j in range(len(A[0])):
            row.append(A[i][j])
        row.append(B[i])
        M.append(row)

    R=rref(M)
    X=[]
    for i in range(len(M)):
        X.append(R[i][-1])
    return X
```

```python
A = [[2, 1], [1, -1]]
B = [5, 1]

print(solve(A, B))
```
