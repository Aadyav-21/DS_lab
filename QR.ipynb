```python
import numpy as np
```

```python
def qr(A):
    m=len(A)
    n=len(A[0])

    R=np.zeros((n,n))
    Q=[]
    for i in range(m):
        row=[]
        for j in range(n):
            row.append(0.0)
        Q.append(row)

    for j in range(n):
        v=[]
        for i in range(m):
            v.append(A[i][j])
        for k in range(j):
            dot=0
            for i in range(m):
                dot+=Q[i][k]*A[i][j]
            for i in range(m):
                v[i]-=dot*Q[i][k]
        l=0
        for i in range(m):
            l+=v[i]*v[i]
        l=l**0.5
        if l<1e-9:
            raise ValueError("can't divide by 0")

        for  i in range(m):
            Q[i][j]=v[i]/l

    Q=np.array(Q)
    A=np.array(A)
    R=Q.T@A

    return Q,R
```

```python
#calculation of eigen values using QR
```

```python
def eig(A,itr=100):
    A=np.array(A)
    a=A.astype(float).copy()

    for _ in range(itr):
        Q,R=qr(a)
        a=R@Q
    return np.diag(a)
```

```python
A = [
    [1, 1,0],
    [1, 0,1],
    [0, 1,1]
]

Q, R = qr(A)

print("Q:")
print(Q)

print("R:")
print(R)

a=eig(A)
a
```
