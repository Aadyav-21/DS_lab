```python
import numpy as np
```

```python
def lu(A):
    n=len(A)
    L=np.eye(n)
    U=np.zeros((n,n))
        
    for i in range(n):
        for j in range(i,n):
            total=0
            for k in range(i):
                total+=L[i][k]*U[k][j]
            U[i][j]=A[i][j]-total

        for j in range(i,n):
            if i==j:
                L[i][j]==1.0
            else:
                total=0
                for k in range(i):
                    total+=L[j][k]*U[k][i]
                if abs(U[i][i])<1e-9:
                    print("LU Decomposition needs pivoting")
                    break
                    
                L[j][i]=(A[j][i]-total)/U[i][i]
    return L,U
                
```

```python
A = [
    [2, 1, 1],
    [4, 3, 3],
    [8, 7, 9]
]

L, U = lu(A)

print("L:")
for row in L:
    print(row)

print("U:")
for row in U:
    print(row)
```

```python
#forward substitution Ly=b
```

```python
def forward(L,b):
    n=len(L)
    Y=[]
    for i in range(n):
        total=0
        for j in range(i):
            total+=L[i][j]*Y[j]
        y=(b[i]-total)/L[i][i]
        Y.append(y)
    return Y
```

```python
def backward(U,Y):
    n=len(U)
    X=[0.0]*n

    for i in range(n-1,-1,-1):
        total=0
        for j in range(i+1,n):
            total+=U[i][j]*X[j]
        X[i]=(Y[i]-total)/U[i][i]
    return X
```
